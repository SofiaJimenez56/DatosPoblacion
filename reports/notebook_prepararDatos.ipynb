```python
llaves = ['ent', 'con', 'upm', 'n_pro_viv', 'v_sel', 'per']

df_integrado = pd.merge(
    df_hog,
    df_viv,
    on=llaves,
    how='inner',
    suffixes=('_hog', '_viv')
)

print("Hogar original:", df_hog.shape)
print("Vivienda original:", df_viv.shape)
print("Integrado:", df_integrado.shape)
print("Columnas esperadas: 132 + 54 - 6 =", 132 + 54 - 6)
```

```python
'''
- El trabajo infantil se mide a nivel hogar en la base de Hogar (`tcti`).
- Las condiciones materiales de la vivienda se miden en la base de Vivienda (`p21`, `p24`, `p25`...).
- La ENTI permite cruzar ambas bases mediante llaves comunes.
'''
```

```python
columnas_relevantes = [
    # Llaves (compartidas)
    'ent', 'con', 'upm', 'n_pro_viv', 'v_sel', 'per',
    # De Hogar
    'tcti', 'tonp', 'tocup', 'tqdna', 'con5a17',
    'escol_jefe', 'cs_p13_2', 'sex_jef', 'clase_hog',
    'tam_hog', 'p_tot', 'clase2_jef',
    # De Vivienda (sin sufijo)
    'p21', 'p22', 'p23', 'p24', 'p25', 'p26', 'p28', 'p31',
    'p32_1', 'p32_2', 'p32_9', 'p32_12',
    'tot_hog', 'pobviv', 'ur2'
]

# Verificar que todas existan antes de seleccionar
faltantes = [c for c in columnas_relevantes if c not in df_integrado.columns]
print("Faltantes:", faltantes)

df_proc = df_integrado[columnas_relevantes].copy()
print("Shape del subconjunto:", df_proc.shape)
df_proc.head()
```

```python
mapa_columnas = {
    # Llaves
    'ent': 'entidad',
    'con': 'control',
    'upm': 'upm',
    'n_pro_viv': 'num_vivienda',
    'v_sel': 'vivienda_sel',
    'per': 'periodo',
    # Hogar
    'tcti': 'trabajo_infantil_total',
    'tonp': 'menores_ocupacion_no_permitida',
    'tocup': 'menores_ocupados_5a17',
    'tqdna': 'menores_quehaceres_no_adecuados',
    'con5a17': 'poblacion_5a17',
    'escol_jefe': 'escolaridad_jefe',
    'cs_p13_2': 'anios_aprobados_jefe',
    'sex_jef': 'sexo_jefe',
    'clase_hog': 'clase_hogar',
    'tam_hog': 'tamano_hogar',
    'p_tot': 'total_residentes',
    'clase2_jef': 'actividad_jefe',
    # Vivienda
    'p21': 'material_piso',
    'p22': 'cuartos_dormir',
    'p23': 'cuartos_totales',
    'p24': 'luz_electrica',
    'p25': 'agua_entubada',
    'p26': 'origen_agua',
    'p28': 'sanitario',
    'p31': 'drenaje',
    'p32_1': 'refrigerador',
    'p32_2': 'lavadora',
    'p32_9': 'computadora',
    'p32_12': 'internet',
    'tot_hog': 'hogares_en_vivienda',
    'pobviv': 'poblacion_vivienda',
    'ur2': 'ambito',
}

df_proc = df_proc.rename(columns=mapa_columnas)
print("Columnas finales:", df_proc.columns.tolist())
print("\nShape:", df_proc.shape)
df_proc.head()
```

```python
cols_num = [
    'trabajo_infantil_total', 'menores_ocupacion_no_permitida',
    'menores_ocupados_5a17', 'menores_quehaceres_no_adecuados',
    'poblacion_5a17', 'escolaridad_jefe', 'anios_aprobados_jefe',
    'sexo_jefe', 'clase_hogar', 'tamano_hogar', 'total_residentes',
    'actividad_jefe', 'material_piso', 'cuartos_dormir', 'cuartos_totales',
    'luz_electrica', 'agua_entubada', 'origen_agua', 'sanitario',
    'drenaje', 'refrigerador', 'lavadora', 'computadora', 'internet',
    'hogares_en_vivienda', 'poblacion_vivienda', 'ambito',
    'tiene_trabajo_infantil',
]

for col in cols_num:
    if col in df_proc.columns:
        df_proc[col] = pd.to_numeric(df_proc[col], errors='coerce')

# Llaves como texto
for col in ['entidad', 'control', 'upm', 'num_vivienda', 'vivienda_sel', 'periodo']:
    if col in df_proc.columns:
        df_proc[col] = df_proc[col].astype(str)

print("Tipos corregidos:")
print(df_proc.dtypes.value_counts())
```

```python
ti_por_escol = df_proc.groupby('escolaridad_jefe')['tiene_trabajo_infantil'].agg(['mean', 'count'])
ti_por_escol['mean'] = (ti_por_escol['mean'] * 100).round(2)
ti_por_escol.columns = ['%_TI', 'n_hogares']

# Añadir etiquetas legibles
etiquetas_escol = {
    1: 'Sin instrucción',
    2: 'Primaria incompleta',
    3: 'Primaria completa',
    4: 'Secundaria',
    5: 'Media superior',
    6: 'Superior',
}
ti_por_escol['nivel'] = ti_por_escol.index.map(etiquetas_escol)
print(ti_por_escol[['nivel', '%_TI', 'n_hogares']])
```

```python
for col, etiquetas in [
    ('luz_electrica', {1: 'Sí', 2: 'No'}),
    ('agua_entubada', {1: 'Sí', 2: 'No'}),
    ('drenaje', None),
    ('internet', {1: 'Sí', 2: 'No'}),
    ('computadora', {1: 'Sí', 2: 'No'}),
]:
    print(f"\n=== {col} ===")
    tabla = df_proc.groupby(col)['tiene_trabajo_infantil'].agg(['mean', 'count'])
    tabla['mean'] = (tabla['mean'] * 100).round(2)
    tabla.columns = ['%_TI', 'n_hogares']
    print(tabla)
```

```python
print("=== Ámbito (1=Urbano, 2=Rural) ===")
tabla = df_proc.groupby('ambito')['tiene_trabajo_infantil'].agg(['mean', 'count'])
tabla['mean'] = (tabla['mean'] * 100).round(2)
tabla.columns = ['%_TI', 'n_hogares']
print(tabla)
```

```python
print("=== Tamaño del hogar ===")
tabla = df_proc.groupby('tamano_hogar')['tiene_trabajo_infantil'].agg(['mean', 'count'])
tabla['mean'] = (tabla['mean'] * 100).round(2)
tabla.columns = ['%_TI', 'n_hogares']
print(tabla)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df_proc[['tiene_trabajo_infantil', 'escolaridad_jefe', 'anios_aprobados_jefe',
                'tamano_hogar', 'total_residentes', 'cuartos_dormir',
                'luz_electrica', 'agua_entubada', 'drenaje', 'internet',
                'computadora']].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.2f')
plt.title('Correlación: trabajo infantil y condiciones del hogar/vivienda')
plt.tight_layout()
plt.show()
```

```python
#Conclusiones 
'''
- El dataset procesado contiene 58,244 hogares con información completa de hogar y vivienda.
- La proporción de hogares con trabajo infantil es de 16.35%
- Aproximadamente 1 de cada 6 hogares tiene un menor en condiciones de trabajo infantil.
- La mayoría de los hogares con TI tienen solo 1 menor afectado, pero hay casos con hasta 5. Los hogares con 3+ menores en TI podrían ser outliers.
- Los estados con mayor TI son Coahuila, BCS y Campeche
- El tamaño del hogar se asocia a una mayor presencia de TI.
- La escolaridad del jefe no es un factor relevante
- Existe una relación amplia entre los servicios que tiene una familia, quien tiene luz, tiene agua y drenaje.
'''
```
